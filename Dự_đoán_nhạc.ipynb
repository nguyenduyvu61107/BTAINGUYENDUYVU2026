```python

!pip install gradio -q
import pandas as pd
import numpy as np
import gradio as gr
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split

data = pd.read_csv('/content/music_data.csv')

le_platform = LabelEncoder()
le_habit = LabelEncoder()
le_genre = LabelEncoder()
data['Nen_tang'] = le_platform.fit_transform(data['Nen_tang'])
data['Thoi_quen'] = le_habit.fit_transform(data['Thoi_quen'])
data['Loai_nhac'] = le_genre.fit_transform(data['Loai_nhac'])

X = data[['Tuoi', 'Thoi_gian_nghe', 'Nen_tang', 'Thoi_quen', 'Muc_do']]
y = data['Loai_nhac']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
model = MLPClassifier(
    hidden_layer_sizes=(32, 16, 8),
    activation='relu',
    solver='adam',
    max_iter=3000,
    random_state=42
)
model.fit(X_scaled, y)
def predict_music(tuoi, thoi_gian, nen_tang, thoi_quen, muc_do):

    if thoi_gian == 0:
        return " Bạn có nghe nhạc đ đâu mà bảo tôi dự đoán?🤔"
    try:
        p_idx = le_platform.transform([nen_tang])[0]
        h_idx = le_habit.transform([thoi_quen])[0]
        input_data = [[tuoi, thoi_gian, p_idx, h_idx, muc_do]]
        input_scaled = scaler.transform(input_data)
        pred_id = model.predict(input_scaled)
        genre_name = le_genre.inverse_transform(pred_id)[0]
        return f"🎧 Thể loại nhạc phù hợp: {genre_name.upper()}"
    except Exception as e:
        return f"Lỗi xử lý: {str(e)}"

custom_css = """
body { background-color: #121212; }
.gradio-container { background-color: #121212 !important; color: white !important; }
button.primary {
    background: linear-gradient(90deg, #FF8C00, #FF4500) !important;
    border: none !important;
    color: white !important;
}
.label-text { color: #FFA500 !important; } /* Màu cam cho label */
"""

with gr.Blocks(theme=gr.themes.Soft(primary_hue="orange", neutral_hue="slate"), css=custom_css) as demo:
    gr.Markdown(
        """
        # <center>VERTA-MAGIC MUSIC PREDICTOR</center>
        <center>Khám phá gu âm nhạc ẩn giấu bên trong bạn .</center>
        """
    )

    with gr.Row():
        with gr.Column():
            gr.Markdown("### 📝 Thông tin cá nhân")
            tuoi = gr.Slider(minimum=10, maximum=80, value=20, step=1, label="Độ tuổi của bạn")
            thoi_gian = gr.Slider(minimum=0, maximum=24, step=0.5, label="Số giờ nghe nhạc mỗi ngày", value=2)

            nen_tang = gr.Dropdown(
                choices=list(le_platform.classes_),
                label="Nền tảng nghe nhạc yêu thích",
                value=list(le_platform.classes_)[0]
            )
            thoi_quen = gr.Dropdown(
                choices=list(le_habit.classes_),
                label="Thói quen khi nghe nhạc",
                value=list(le_habit.classes_)[0]
            )
            muc_do = gr.Radio(
                choices=[1, 2, 3],
                label="Mức độ hưng phấn (1: Bình thường, 2:Hưng phấn, 3: Nhún nhảy)",
                value=2
            )

            btn = gr.Button("PHÂN TÍCH NGAY GU NHẠC CỦA BẠN🔥 ", variant="primary")

        with gr.Column():
            gr.Markdown("### Kết quả phân tích")
            output = gr.Textbox(label="Dự đoán", placeholder="Kết quả sẽ hiển thị tại đây...")

            gr.Image("https://media.istockphoto.com/id/1391884768/vi/vec-to/bu%E1%BB%95i-h%C3%B2a-nh%E1%BA%A1c-c%E1%BB%A7a-ban-nh%E1%BA%A1c-alternative-v%E1%BB%9Bi-crowd-silhouettes.jpg?s=612x612&w=0&k=20&c=ZS3nuWbRMmUxTouqmYtssXO9hD_qhMq56r2GnPbbLWc=",
                     label="Music Vibe", show_label=False)

    btn.click(
        fn=predict_music,
        inputs=[tuoi, thoi_gian, nen_tang, thoi_quen, muc_do],
        outputs=output
    )



demo.launch(share=True)
```
